```python
queue = []
queue.append(10)
queue.append(20)
queue.append(30)

print(queue)

queue.pop(0)
print(queue)
```

```python
from collections import deque
queue = deque()

queue.append(10)
queue.append(20)
queue.append(30)
queue.append(40)

print("Queue ", queue)

removed = queue.popleft()
print("Removed ", removed)
print("After removing, queue ", queue)

print("Front ", queue[0])
print("Rear ", queue[-1])
print("Length ", len(queue))
```

```python
##print number from 1 to 10 using recurrsion

def print_n(n):
    if n == 1:
        print(1)
        return
    print_n(n - 1)
    print(n)

print_n(10)
```

```python
def sum_n(n):
    if (n == 1):
        return 1
    return n + sum_n(n-1)
print(sum_n(10))
```
