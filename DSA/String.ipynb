```python
##Count vowels

count = 0
abc = "aeiouAEIOU"

str = "I study in Lovely Professional University"

for i in str:
    if i in abc:
        count = count + 1
print(count)
```

```python
##Reverse a string

str = "I study in Lovely Professional University"
print(str[::-1])
reverse_str = ""
for i in str:
    reverse_str = i+reverse_str
print(reverse_str)
```

```python
##Check a string is palindrome or not

str = input()
new_str = str

reverse_str = ""
for i in str:
  reverse_str = i + reverse_str

if new_str == reverse_str:
    print("Palindrome string")
else:
    print("Not Palindrome")
```

```python
##Frequency of each character using dictionary


str = input()
for i in str:
    count = 0
    for j in str:
        if i == j:
            count += 1
    print(i, count)
```

```python
##First occurence in a string

def first_occurrence(s, ch):
    for i in range(len(s)):
        if s[i] == ch:
            return i
    return -1

# Example
s = "programming"
ch = "g"

print(first_occurrence(s, ch))
```
