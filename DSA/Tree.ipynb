```python
class treeNode:
    def __init__(self,value):
        self.value=value
        self.left=None
        self.right=None

root=treeNode(1)
root.left=treeNode(2)
root.right=treeNode(3)
root.left.left=treeNode(4)
root.left.right=treeNode(5)
root.right.left=treeNode(6)
root.right.right=treeNode(7)
print("Tree created successfully!")
```

```python
def preorder(node):
    if node:
        print(node.value,end=" ")
        preorder(node.left)
        preorder(node.right)

print("preorder")
preorder(root)
```

```python
def inorder(node):
    if node:
        preorder(node.left)
        print(node.value,end=" ")
        preorder(node.right)

print("inorder")
inorder(root)
```

```python
def postorder(node):
    if node:
        preorder(node.left)
        preorder(node.right)
        print(node.value,end=" ")

print("postorder")
postorder(root)
```

```python
from collections import deque
def level_order(root):
    if not root:
        return
    q=deque([root])
    while q:
        node = q.popleft()
        print(node.value,end=" ")
        if node.left:
            q.append(node.left)
        if node.right:
            q.append(node.right)
print("Level Order")
level_order(root)
```

```python
def height(node):
    if node is None:
        return -1

    return 1+max(height(node.left),height(node.right))

print("height: ",height(root))
```

```python
def count_nodes(node):
    if node is None:
        return 0

    return 1+count_nodes(node.left)+count_nodes(node.right)

print("Node: ",count_nodes(root))
```

```python
def count_leaf_nodes(node):
    if node is None:
        return 0
    if node.left is None and node.right is None:
        return 1
    return count_leaf_nodes(node.left)+count_leaf_nodes(node.right)

print("Count of Leaf Nodes: ",count_leaf_nodes(root))
```
