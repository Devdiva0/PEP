```python
##Stack

stack = []
stack.append(10)
stack.append(20)
stack.append(30)

print("After push ", stack)

##Pop

print("Popped Element", stack.pop())

print("After pop ", stack)
```

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None
```

```python
class Stack:
    def __init__(self):
        self.top = None
        self.count = 0   # Stores the number of elements

    # Push operation
    def push(self, data):
        new_node = Node(data)
        new_node.next = self.top
        self.top = new_node
        self.count += 1

    # Pop operation
    def pop(self):
        if self.is_empty():
            print("Stack Underflow")
            return None

        popped = self.top.data
        self.top = self.top.next
        self.count -= 1
        return popped

    # Peek operation
    def peek(self):
        if self.is_empty():
            return None
        return self.top.data

    # Check if stack is empty
    def is_empty(self):
        return self.top is None

    # Return length of stack
    def length(self):
        return self.count

    # Display stack
    def display(self):
        temp = self.top
        while temp:
            print(temp.data, end=" -> ")
            temp = temp.next
        print("None")
```

```python
s = Stack()

s.push(10)
s.push(20)
s.push(30)

print("Length:", s.length())   # 3

s.pop()

print("Length:", s.length())   # 2
```
