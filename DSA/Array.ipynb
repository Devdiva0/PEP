```python
# sum of maxi and mini elements
arr=[12,53,92,4]
maxi=arr[0]
for i in arr:
    if i>maxi:
        maxi=i

mini=arr[0]
for i in arr:
    if i<mini:
        mini=i

sum_max_min=maxi+mini
print(sum_max_min)
```
