```python
# sort vs sorted
# sort()	                        sorted()
# Method of a list	                Built-in function
# Sorts the original list	        Returns a new sorted list
# Changes the original list	        Original data remains unchanged
# Returns None	                    Returns the sorted list
# Works only on lists	            Works on any iterable (list, tuple, string, set, etc.)
```

```python
numbers = [4, 2, 5, 1]

numbers.sort()

print(numbers)
```

```python
numbers = [4, 2, 5, 1]

new_list = sorted(numbers)

print(numbers)
print(new_list)
```

```python
# reverse vs reversed
# reverse()             	         reversed()
# List method	                     Built-in function
# Reverses the original list	     Returns a reverse iterator
# Returns None	                     Returns an iterator
# Works only on lists	             Works on lists, tuples, strings, etc.
```

```python
numbers = [1, 2, 3, 4]

numbers.reverse()

print(numbers)
```

```python
numbers = [1, 2, 3, 4]

rev = reversed(numbers)

print(list(rev))
print(numbers)
```
