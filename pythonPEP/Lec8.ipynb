```python
#factorial
def fact(n):
    if n==0:
        return 1
    return n*fact(n-1)
    
fact(5)
```

```python
#sum of n nums
def sum(n):
    if n==0:
        return 0
    return n+sum(n-1)
sum(6)
```

```python
def summ(nums,n):
    if n==0:
        return nums[0]
    return nums[n]+summ(nums,n-1)

nums=[12,5,7,10,8]
n=len(nums)-1
print(summ(nums,n))
```

```python
#recursive function to find max number in nested list [10,[25,6],[7,[55,2],11],30] ->55
def maximum(lis,n):
    if n==0:
        return lis[n]
    return lis[n]>lis[n-1]
```

```python
#compress 'aaabbbccccdda' to 'a3b3c4d2a1' using recursive only.
def compress(s, index=0, count=1):
    # Base case
    if index == len(s) - 1:
        return s[index] + str(count)

    # Same character
    if s[index] == s[index + 1]:
        return compress(s, index + 1, count + 1)

    # Different character
    return s[index] + str(count) + compress(s, index + 1, 1)


s = "aaabbbccccdda"
print(compress(s))
```

```python
#convert 5072 into 'Five Zero Seven Two' using recursive only.
digits = ["Zero", "One", "Two", "Three", "Four",
          "Five", "Six", "Seven", "Eight", "Nine"]

def number_to_words(num):
    if len(num) == 0:
        return

    print(digits[int(num[0])], end=" ")
    number_to_words(num[1:])

number = "5072"
number_to_words(number)
```

```python
#Implement my_reduce(function,data) without using functools.reduce(). example: lambda x,y:x+y on [2,4,6,8]->20
def my_reduce(func, data):
    result = data[0]      # Start with the first element

    for i in range(1, len(data)):
        result = func(result, data[i])

    return result


data = [2, 4, 6, 8]

ans = my_reduce(lambda x, y: x + y, data)

print(ans)
```

```python
#convert a decimal number to binary using recursion only. Do not use bin().
def decimal_to_binary(n):
    if n == 0:
        return

    decimal_to_binary(n // 2)
    print(n % 2, end="")


num = 13

if num == 0:
    print(0)
else:
    decimal_to_binary(num)
```

```python
# Given employee salary tuples:
# (1) increase every salary 10% using map(),
# (2) Filter salaries > ₹60000 using Filter(),
# (3) find total salary using rduce()

from functools import reduce

salaries = (45000, 55000, 60000, 75000, 80000)

# (1) Increase every salary by 10%
increased = list(map(lambda x: x * 1.10, salaries))
print("Increased Salaries:", increased)

# (2) Filter salaries > 60000
filtered = list(filter(lambda x: x > 60000, increased))
print("Salaries > 60000:", filtered)

# (3) Find total salary
total = reduce(lambda x, y: x + y, increased)
print("Total Salary:", total)
```
