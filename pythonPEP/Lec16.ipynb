```python
class students:
    def __init__(self,rollno,name,marks):
        self.name=name
        self.rollno=rollno
        self.marks=marks

    def get_grade(self):
        if self.marks>=90:
            print("Grade: A")
        elif self.marks >=75:
            print("Grade: B")
        else:
            print("Grade: C")

s1=students(1,"Dev",66)
s1.get_grade()

        
```

```python
class employee:
    def __init__(self,name,salary):
        self.name=name
        self.__salary=salary

    @property
    def salary(self):
        return self.__salary
        
    @salary.setter
    def salary(self,value):
        if value>0:
            self.__salary=value
        else:
            print("Salary must be positive")

    def apply_raise(self,percent):
        self.__salary+=self.__salary*(percent/100)
        print(f"{percent}% raise applied successfully")

emp=employee("Dev",50000)

print(emp.salary)

emp.apply_raise(20)
print(emp.salary)
    
```

```python
class employee:
    def __init__(self,name,salary):
        self.name=name
        self.__salary=salary

    def get_salary(self):
        return self.__salary
        
    def set_salary(self,value):
        if value>0:
            self.__salary=value
        else:
            print("Salary must be positive")

    def apply_raise(self,percent):
        self.__salary+=self.__salary*(percent/100)
        print(f"{percent}% raise applied successfully")

emp=employee("Dev",50000)

print(emp.get_salary())

emp.apply_raise(20)
print(emp.get_salary())
emp.set_salary(4000)
print(emp.get_salary())
```
