```python
import pandas as pd
import numpy as np
df=pd.read_csv("/Users/devangi.yadavicloud.com/Downloads/student_performance_project.csv")
print(df.shape)
print(df.info)
print(df.head(5))
print("Duplicate rows:",df.duplicated().sum())
```

```python
df.isnull().sum()
```

```python
df["Math"]=df["Math"].fillna(df['Math'].mean().round(2))
print(df.head(10))
```

```python
df["Science"]=df["Science"].fillna(df['Science'].mean().round(2))
print(df.head(10))
```

```python
df["English"]=df["English"].fillna(df['English'].mean().round(2))
print(df.head(10))
df.isnull().sum()
```

```python
df.drop_duplicates(subset=['StudentID','City'], keep='first')
print("Duplicate rows:",df.duplicated().sum())
```

```python
df['total_marks']=df["Math"]+df["Science"]+df["English"]
top_students=df[df['total_marks']>240]
print(top_students[['StudentID', 'Name', 'City','total_marks']])
```

```python
subject_avg = df[['Math', 'Science', 'English']].mean().round(2)
print(subject_avg)
```

```python
df['Avg']=df[["Math","Science","English"]].mean(axis=1).round(2)
df['Rank'] = df['total_marks'].rank(ascending=False, method='min').astype(int)
rank=df.sort_values('Rank')
print(df[['Rank','StudentID','Name','total_marks','Avg']])
```

```python
PASS_MARK = 33
df['Result'] = np.where(
 (df['Math'] >= PASS_MARK) & (df['Science'] >= PASS_MARK) & (df['English'] >= PASS_MARK),
 'Pass', 'Fail'
)
print(df[['StudentID','Name','Math','Science','English','Result']])
```

```python
gender_report = df.groupby('Gender')['Avg'].mean().round(2)
city_report = df.groupby('City')['Avg'].mean().round(2)
print(gender_report)
print(city_report)
```

```python
report=df.groupby(['City','Gender']).agg(
    students=('StudentID', 'count'),
    avg_total=('total_marks', 'mean'),
    pass_rate=('Result', lambda x: (x == 'Pass').mean() * 100)
).round(2)
print(report)
```

```python
topper = df.sort_values('total_marks', ascending=False)[['Rank', 'Name', 'City', 'total_marks',
'Avg', 'Result']]
print(topper.head(5))
```

```python
df.to_csv('student_performance_report.csv', index=False)
print("Sucessfully exported")
```
