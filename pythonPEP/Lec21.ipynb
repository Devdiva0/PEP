```python
import numpy as np

# dataset creation 
student_names = np.array([
    "Aarav",
    "Ananya",
    "Rohan",
    "Priya",
    "Arjun",
    "Sneha",
    "Rahul",
    "Kavya",
    "Vikram",
    "Meera"
])

# Rows = Students
# Columns = Subjects (Maths, Science, English, Computer, Social Science)

students = np.array([
    [85, 78, 92, 88, 76], 
    [67, 72, 70, 75, 68],  
    [90, 95, 89, 94, 91],   
    [55, 60, 58, 62, 57],   
    [80, 85, 84, 82, 81],
    [73, 77, 75, 79, 74],  
    [96, 92, 98, 97, 95],   
    [64, 69, 66, 70, 68],   
    [88, 90, 85, 91, 89],   
    [78, 81, 79, 80, 77]    
])

subjects = ["Maths", "Science", "English", "Computer", "Social Science"]
```

```python
# average marks per student
avg_marks=np.mean(students, axis=1)
for name, avg in zip(student_names, avg_marks):
    print(f"{name}: {avg:.2f}")
```

```python
# highest and lowest marks
highest=np.max(students, axis=1)
lowest=np.min(students, axis=1)
for name, high, low in zip(student_names, highest, lowest):
    print(f"{name} -> Highest: {high}, Lowest: {low}")
```

```python
# standard deviation per student
std1=np.std(students,axis=1)
for name, std1 in zip(student_names,std1):
    print(f"{name} -> std: {std1}")
```

```python
# ranking students
avg=np.mean(students,axis=1)
rank=np.argsort(avg)[::-1]
print("Ranking of Students")
for i, idx in enumerate(rank, start=1):
    print(f"Rank {i}: {student_names[idx]} ({avg[idx]:.2f})")
```
