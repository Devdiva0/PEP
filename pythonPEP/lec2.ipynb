```python
#prime number
num=int(input("Enter number: "))
if num <=1 :
    print("not prime")
else:
    for i in range(2,num):
        if num%i==0:
            print("Not Prime")
            break
    else:
        print("Prime")
```

```python
#take 2 lists with the corresponding pwds
usernames = ["tanvi", "ragini", "amit"]
passwords = [1234, 5678, 9999]
#username input
username_entered = input("Enter username: ")
#check
if username_entered in usernames:
    user_index = usernames.index(username_entered)   #check index of username entered from usernames list
    pwd_entered = int(input("Enter pwd: "))     #pwd input
    if pwd_entered == passwords[user_index]:       #check index of pwds entered from pwds list
        print("correct user")
    else:
        print("incorrect password")
else:
    print("username not found")
```

```python
#withdraw amount from atm code
#list
usernames = ["tanvi", "rahul"]
passwords = [1234, 5678]
balance = [7000.0, 11000.0]
#login
user_entered = input("Enter username: ")
#chk
if user_entered in usernames:
    user_idx = usernames.index(user_entered)            #pos of user in list
#pin input
    pin_entered = int(input("Enter PIN: "))
    
    if pin_entered == passwords[user_idx]:
        print("Login successful!")
#withdraw
        amount = float(input("Enter amount to withdraw: ")) 
        if amount > balance[user_idx]:
            print("Insufficient balance")
        elif amount <= 0:
            print("Invalid amount")
        else:
            print("Withdrawal successful")
    else:
        print("Incorrect pin")
else:
    print("Username not found")
```
