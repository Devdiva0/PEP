```python
n=int(input())
m=n
rev=0
while m>0:
    digit=m%10
    rev=rev*10 + digit
    m=m//10
print(rev)
```

```python
n=int(input())
m=n
rev=0
while m>0:
    digit=m%10
    rev=rev*10 + digit
    m=m//10
if rev==n:
    print("Is Palindrome")

else:
    print("Not Palindrome")
```
