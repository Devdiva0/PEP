#calculator code

```python
num1 = float(input("Enter first number:"))
num2 = float(input("Enter second number:"))
operator = input("Enter operator: ")
if operator == "+":
    print("Result =",num1 + num2)
elif operator == "-":
    print("Result =",num1 - num2)
elif operator == "*":
    print("Result =",num1 * num2)
elif operator == "/":
    if num2 != 0 :
        print("Result =",num1 / num2)
    else: 
        print("div by 0 not allowed")
elif operator == "==":
    print("Result =",num1 == num2)
elif operator == ">":
    print("Result =",num1 > num2)
elif operator == "-=":
    num1 -= num2  
    print("Result =", num1)
else:
    print("invalid number")

```
