<h4>--------------------------------------MAP------------------------------------------------------

```python
def double(x):
    return x*2

nums=[1,2,3,4,5,6]
doubled = map(double,nums)

print(list(doubled))
```

<h4>--------------------------------Filter---------------------------------------

```python
nums=[1,2,3,4,5,6,7,8,44]

filtered=list(filter(lambda x:x%2==0,nums))
print(filtered)
```

<h4>---------------------------------Reduce----------------------------------------

```python
from functools import reduce

nums=[1,2,3,4,5,6,7]

total=reduce(lambda x,y:x+y,nums)
print("Sum",total)
```

```python
#prime
import math
def is_prime(n):
    if n<2:
        return False
    return len(list(filter(lambda x:n%x==0, range(2,int(math.sqrt(n))))))==0

num=4
   
if is_prime(num):
    print("prime")
else:
    print("not prime")
```
