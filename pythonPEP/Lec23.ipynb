```python
import pandas as pd
import numpy as np

df=pd.read_csv("/Users/devangi.yadavicloud.com/Downloads/students_clean.csv")
df.head(10)
```

```python
print(df.groupby('City')['Math'].mean().round(2))
```

```python
print(df.groupby('City')['Math'].agg(['count','min','max','median','std']).round(2))
```

```python
report=df.groupby('City').agg(
    total_math=('Math','sum'),
    avg_math=('Math','mean'),
    students=('Math','count')
).round(2)
print(report)
```

```python
print(df.groupby(['City','Gender'])['Math'].mean().round(2))
```

```python
report=df.groupby(['City','Gender']).agg(
    total_math=('Math','sum'),
    avg_math=('Math','mean'),
    students=('Math','count')
).round(2)
print(report)
```
