```python
#check plindrome
string="HEllEH"
le=len(string)
rev=""
for i in range(0,le):
    rev+=string[le-1]
    le-=1
    
if rev==string:
    print("Plindrome")
else:
    print("Not plindrome")
```

```python
#count the number of letters, digits, spaces, and special characters.
s = "Python123 @ LPU!"

letters = 0
digits = 0
spaces = 0
special = 0

for ch in s:
    if ch.isalpha():
        letters += 1
    elif ch.isdigit():
        digits += 1
    elif ch.isspace():
        spaces += 1
    else:
        special += 1

print("Letters :", letters)
print("Digits :", digits)
print("Spaces :", spaces)
print("Special Characters :", special)
```

```python
#Find the longest word in a sentence
sent="heloo, helooo my name is Devvvvvanngggiiiiiii , heheheh"
words=sent.split()
longest=""
for word in words:
    if len(word)>len(longest):
        longest=word
print(longest)
```
