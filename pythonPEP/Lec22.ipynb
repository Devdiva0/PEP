```python
data = {
    'Name': ['Aditi', 'Rohan', 'Meera'],
    'Age': [21, 23, 22],
    'City': ['Delhi', 'Mumbai', 'Pune']
}

df = pd.DataFrame(data)
#1. Age greater than 22
print(df[df['Age'] > 22])
#2. Age > 22 AND City is Mumbai
print(df[(df['Age'] > 22) & (df['City'] == 'Mumbai')])
#3. Age > 22 OR City is Delhi
print(df[(df['Age'] > 22) | (df['City'] == 'Delhi')])
#4. City is not Delhi
print(df[df['City'] != 'Delhi'])
```

```python
data = {
    'Name': ['Aditi', 'Rohan', 'Meera'],
    'Age': [21, 23, 22],
    'Marks': [85, np.nan, 90],
    'City': ['Delhi', 'Mumbai', 'Pune']
}

df = pd.DataFrame(data)
df.isnull()             #boolean mask,true where missing
df.isnull().sum()       #count of missing values per col
df.notnull()

print(df.dropna())
print(df.dropna(subset=['Marks']))
print(df.dropna(how='all'))

df['Marks'] = df['Marks'].fillna(0)
print(df)
df['Marks'] = df['Marks'].fillna(df['Marks'].mean())
df.fillna(method='ffill')

df['City'] = df['City'].replace({
    'Delhi': 'New Delhi',
    'Pune': 'Pune City'
})

print(df)

df = df.rename(columns={
    'Name': 'Student_Name',
    'Age': 'Student_Age'
})

print(df)

print(df.duplicated())

print(df.drop_duplicates())
df = pd.DataFrame({
    'Student_Name': ['Aditi', 'Rohan', 'Meera', 'Aditi'],
    'Student_Age': [21, 23, 22, 25],
    'City': ['Delhi', 'Mumbai', 'Pune', 'Delhi']
})
print(df)

#Keep the first "Aditi":
print(df.drop_duplicates(subset=['Student_Name'], keep='first'))

#Keep the last "Aditi":
print(df.drop_duplicates(subset=['Student_Name'], keep='last'))
```
